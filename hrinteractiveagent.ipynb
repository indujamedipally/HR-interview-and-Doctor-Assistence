```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver
from langchain_google_genai import ChatGoogleGenerativeAI
from google.colab import userdata
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=userdata.get("YOUR_GOOGLE_API_KEY")
)
```

```python
hr_prompt = """
You are a professional HR interviewer preparing a candidate for a job interview.
Your goal is to practice interview questions with the user.
1. Start by asking for their name and the role they are interviewing for and after that only 2-3 questions.
2. Ask one interview question at a time.
3. Provide constructive feedback on their answers before moving to the next question.
4. Keep the conversation professional and encouraging but after the interview provide them what was a mistake in interview response and what was blunder.
"""
```

```python
agent = create_agent(
    model=llm,
    tools=[],
    system_prompt=hr_prompt,
    checkpointer=InMemorySaver(),
)
```

```python
from langchain_core.messages import HumanMessage
config = {"configurable": {"thread_id": str(uuid7())}}
print("HR Agent: Hello! I'm here to help you prepare for your interview.")
first_turn = True
while True:
    if first_turn:
        user_input = input("\nTo start, please enter your name and the role you are interviewing for: ")
        first_turn = False
    else:
        user_input = input("\nYour answer (or type 'End' to stop): ")
    if user_input.strip().lower() == "end":
        break
    result = agent.invoke(
        {"messages": [HumanMessage(content=user_input)]},
        config=config,
    )
    print(f"\nHR Agent: {result['messages'][-1].content}")
print("\nThanks for practicing! Good luck with your interview.")
```

```python
doc_prompt ="""
You are a professional Doctor treating a patient for their diesease,
Your goal is to practice questions with the user regarding their health
1.Start by asking their name and after that ask only 2 questions.
2.Ask one question at a time.
3.provide constructive feedback on their answers before moving to next question
4.Keep the conversation professional and encouraging but after the questions provide them what are precautions they need to take in response.
"""
```

```python
agent = create_agent(
    model=llm,
    tools=[],
    system_prompt=doc_prompt,
      checkpointer=InMemorySaver(),
)
from langchain_core.messages import HumanMessage


config = {"configurable": {"thread_id": str(uuid7())}}


print("HR Agent: Hello! I'm here to help you Treat you")


first_turn = True


while True:
    if first_turn:
        user_input = input("\nTo start, please enter your name ")
        first_turn = False
    else:
        user_input = input("\nYour answer (or type 'End' to stop): ")


    if user_input.strip().lower() == "end":
        break


    result = agent.invoke(
        {"messages": [HumanMessage(content=user_input)]},
        config=config,
    )


    print(f"\nDoc Agent: {result['messages'][-1].content}")


print("\nSpeed Recovery")
```
